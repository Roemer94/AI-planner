## 2. What are Tensors?

In machine learning and deep learning, all data is ultimately stored as numbers. To efficiently organize and process these numbers, we use **tensors**.

A tensor is best thought of as a **multi-dimensional array** - an extension of things you already know:
- A single number (e.g., 5) is a tensor of dimension 0
- A list (e.g., [1,2,3]) is a tensor of dimension 1
- A table (matrix) is a tensor of dimension 2
- Higher dimensions occur when you store multiple images at once

```python
#!pip install numpy
import numpy as np

# Tensor rank 0: scalar
scalar = np.array(5)
print("Rank 0 (scalar):", scalar)

# Tensor rank 1: vector
vector = np.array([1, 2, 3, 4, 5])
print("\nRank 1 (vector):", vector)

# Tensor rank 2: matrix
matrix = np.array([[1, 2, 3],
                    [4, 5, 6]])
print("\nRank 2 (matrix):\n", matrix)
```

## 3. Shape of a Tensor

## 4. Rank (Number of Dimensions)

The **rank** of a tensor is the number of dimensions it has. It shows how "complex" the structure is:
- Rank 0: scalar (one value)
- Rank 1: vector (list)
- Rank 2: matrix (table)
- Rank 3+: higher dimensions

```python
# Comparing rank and shape
scalar = np.array(5)
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2, 3], [4, 5, 6]])
tensor_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("Scalar - Rank:", scalar.ndim, ", Shape:", scalar.shape)
print("Vector - Rank:", vector.ndim, ", Shape:", vector.shape)
print("Matrix - Rank:", matrix.ndim, ", Shape:", matrix.shape)
print("3D Tensor - Rank:", tensor_3d.ndim, ", Shape:", tensor_3d.shape)
```

## 5. Size of a Tensor

The **size** of a tensor is the total number of elements. Calculate it by multiplying all dimensions together.

```python
# Calculating tensor size
matrix = np.array([[1, 2, 3],
                    [4, 5, 6]])

print("Matrix shape:", matrix.shape)
print("Total size: 2 × 3 =", matrix.size)

tensor_3d = np.random.rand(3, 4, 2)
print("\n3D Tensor shape:", tensor_3d.shape)
print("Total size: 3 × 4 × 2 =", tensor_3d.size)
```

## Making and Improving Predictions

A model starts by making a **prediction** and comparing it to the **true value**.  
The difference is measured with a loss function.

Example target rule:

**output** = 2 × **input**

Suppose:
- input = 2
- true output = 4

If the model currently guesses **output** = 1 × **input**, then:

**ŷ** = 1 × 2 = 2

Using squared error:

**loss** = (ŷ - y)² = (2 - 4)² = 4

---

## Goal: Minimize the Loss

Training tries to make this loss as small as possible by updating model parameters.

---

## Gradient Descent (Intuition)

Gradient descent is a step-by-step optimization method:

1. Compute the slope (gradient) of the loss.
2. Move a small step in the opposite direction.
3. Repeat.

Update rule:

**x_new** = **x** - α · **f'**(**x**)

where α is the learning rate.

---

## Example: Minimize **f(x) = x²**

- **f'**(**x**) = 2**x**
- Start at **x = 3**, α = 0.1

Step 1:

**x** = 3 - 0.1 · 6 = 2.4

Step 2:

**x** = 2.4 - 0.1 · 4.8 = 1.92

So **x** moves toward 0, the minimum of **x²**.

---

## 1. Load a Pretrained Language Model

We load a small pretrained causal language model and its tokenizer. This model is already trained and can be used immediately for text completion.

A **causal language model** predicts the next token in a sequence from left to right, based on previous tokens. Think of it as reading a sentence word by word and estimating what comes next—just like the exercises in the PDF.

```python
# this install may take a while, but it only needs to be done once. after that, you can comment it out
#%pip install transformers
import torch
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForCausalLM

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_name = "distilgpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(model_name).to(device)
model.eval()

print("Device:", device)
print("Model loaded:", model_name)
print("Vocabulary size:", tokenizer.vocab_size)
```

Some tests to check if the model is working:

```python
# Test the model and tokenizer
test_text = "Hello, how are"
inputs = tokenizer(test_text, return_tensors="pt").to(device)
print("\nTest tokenization:")
print(f"Input text: '{test_text}'")
print(f"Token IDs: {inputs['input_ids']}")
print(f"Tokens: {tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])}")

# Generate next token prediction
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits
    next_token_logits = logits[0, -1, :]
    next_token_id = next_token_logits.argmax().item()
    next_token = tokenizer.decode([next_token_id])
    print(f"\nPredicted next token: '{next_token}'")
```

## Tokenize a prompt: 

We take a short prompt and convert it into tokens that the model can understand. This is done using the tokenizer associated with the pretrained model. The tokenizer breaks down the input text into smaller pieces (tokens) and converts them into numerical format that the model can process. Think of this as translating human language into a format that the AI can understand and work with. Like how we break down sentences into words and letters to understand them better, the tokenizer does something similar for the AI model.

```python
prompt = "He was a large, heavy man with a very big"

encoded = tokenizer(prompt, return_tensors="pt")
input_ids = encoded["input_ids"].to(device)
attention_mask = encoded["attention_mask"].to(device)

tokens = tokenizer.convert_ids_to_tokens(input_ids[0].tolist())

print("Prompt:")
print(prompt)
print()
print("Token ids:")
print(input_ids[0].tolist())
print()
print("Tokens:")
print(tokens)
```

# 3. Predict the Next Token

Next, we give the model a prompt and let it predict the next token.

This is the core idea behind language models: they read a sequence of tokens and estimate what comes next based on patterns learned during training. Different prompts produce different probability distributions and therefore different continuations.

```python
with torch.no_grad():
    outputs = model(input_ids=input_ids, attention_mask=attention_mask)
    next_token_logits = outputs.logits[0, -1, :]
    next_token_probs = torch.softmax(next_token_logits, dim=-1)

top_k = 10
top_probs, top_ids = torch.topk(next_token_probs, k=top_k)

top_tokens = tokenizer.convert_ids_to_tokens(top_ids.tolist())
top_words = [tokenizer.decode([tid]) for tid in top_ids.tolist()]

print(f"Top next-token predictions for the sentence '{prompt}...':")
for rank, (tok, word, prob) in enumerate(zip(top_tokens, top_words, top_probs.tolist()), start=1):
    print(f"{rank:>2}. token={tok:<15} text={word!r:<18} prob={prob:.4f}")
```

Sometimes two or more next tokens have very similar probabilities. In that case, there may not be one clearly "best" continuation.

This is normal in language modeling: natural language is often ambiguous, and multiple continuations can be valid in context.

A useful metric here is **perplexity**. Perplexity measures how well a model predicts a sequence, and is defined as:

$$
\text{Perplexity} = \exp\left(\text{average negative log-likelihood}\right)
$$

- Lower perplexity means the model assigns higher probability to the true next tokens.
- Higher perplexity means the model is more "surprised" by the text.
- Perplexity values should only be compared on the same dataset and tokenization setup.

So, ambiguity is a property of language/context, while perplexity is a quantitative model-evaluation metric.

# Comparing Candidate Next Words

Now we get to an interesting part: we will observe the model's ability to make a **decision** between two or more candidate next words.

This is where context matters. The model assigns a probability to each candidate continuation and we can compare those probabilities directly. By looking at these scores, you can see how the model interprets the prompt and which continuation it considers most likely.

```python
def continuation_logprob(prompt_text, continuation_text):
    full_text = prompt_text + continuation_text
    prompt_ids = tokenizer(prompt_text, return_tensors="pt")["input_ids"][0].to(device)
    full_ids = tokenizer(full_text, return_tensors="pt")["input_ids"][0].to(device)

    with torch.no_grad():
        out = model(full_ids.unsqueeze(0))
        log_probs = torch.log_softmax(out.logits[0], dim=-1)

    start = len(prompt_ids)
    total = 0.0

    for pos in range(start, len(full_ids)):
        prev_pos = pos - 1
        token_id = full_ids[pos].item()
        total += log_probs[prev_pos, token_id].item()

    return total

candidates = [
    " moustache",
    " beard",
    " hat",
    " neck",
]

scores = []
for cand in candidates:
    score = continuation_logprob(prompt, cand)
    scores.append(score)

scores_tensor = torch.tensor(scores)
cand_probs = torch.softmax(scores_tensor, dim=0)

print("Candidate comparison for the same prompt:")
for cand, score, prob in zip(candidates, scores, cand_probs.tolist()):
    print(f"{cand!r:<14} logprob={score:.3f}   relative_prob={prob:.3f}")
```

`logprob` (log probability) is the natural logarithm of probability. We use log probabilities because they are numerically stable when probabilities are very small.

In language modeling, sequence probabilities are products of many token probabilities. In log space, multiplication becomes addition, which is easier and more stable:

\[
\log(p_1 p_2 \cdots p_n) = \log p_1 + \log p_2 + \cdots + \log p_n
\]

Use this section to experiment with prompts and compare candidate continuations by their log probabilities. When log probabilities are close, the model is less certain. When one candidate has a much higher log probability, the model is more confident.

This uncertainty is one reason models can make mistakes or hallucinate, especially on ambiguous or low-information prompts.

## 7. Demonstrate Context Sensitivity

A language model is highly sensitive to context.
If we slightly change the prompt, the next-token preferences also change.

```python
prompts = [
    "The cat sat on the",
    "The cat sat under the",
]

for text in prompts:
    enc = tokenizer(text, return_tensors="pt")
    enc = {k: v.to(device) for k, v in enc.items()}

    with torch.no_grad():
        out = model(**enc)
        probs = torch.softmax(out.logits[0, -1, :], dim=-1)
        top_p, top_i = torch.topk(probs, k=5)

    print("=" * 70)
    print("Prompt:", text)
    print("Top predictions:")
    for rank, (tid, prob) in enumerate(zip(top_i.tolist(), top_p.tolist()), start=1):
        print(f"{rank:>2}. {tokenizer.decode([tid])!r:<16} prob={prob:.4f}")
```

Here we observe the influence of the word "under" on the model's next-token predictions.

A single preposition can change the model's expectations for what should come next. This demonstrates how strongly context affects language-model behavior.

# 8. N-gram Analysis

Here is a small handmade mini language-model table, so you can show the idea behind next-word prediction without extra training.

```python
import random

manual_table = {
    ("Hij",): {"was": 0.7, "had": 0.2, "zag": 0.1},
    ("was",): {"een": 0.8, "de": 0.2},
    ("een",): {"grote": 0.5, "kleine": 0.3, "vriendelijke": 0.2},
    ("grote",): {"man": 0.7, "snor": 0.3},
    ("man",): {"met": 0.8, "en": 0.2},
    ("met",): {"een": 0.9, "zijn": 0.1},
    ("een", "grote"): {"man": 0.7, "snor": 0.3},
}

def next_word_distribution(history):
    key = tuple(history[-1:])
    if key in manual_table:
        return manual_table[key]
    return {"<eos>": 1.0}

def sample_from_distribution(dist):
    words = list(dist.keys())
    probs = list(dist.values())
    return random.choices(words, weights=probs, k=1)[0]

def manual_generate(start_words, max_steps=8):
    words = list(start_words)
    for _ in range(max_steps):
        dist = next_word_distribution(words)
        nxt = sample_from_distribution(dist)
        if nxt == "<eos>":
            break
        words.append(nxt)
    return " ".join(words)

history = ["Hij"]
dist = next_word_distribution(history)

print("History:", history)
print("Next-word distribution:")
for word, prob in dist.items():
    print(f"{word:<12} {prob:.2f}")

print()
print("Three sample continuations:")
for _ in range(3):
    print("-", manual_generate(["Hij"], max_steps=6))
```
